```python
import numpy as np
import pandas as pd
```

```python
data = pd.read_csv("/content/diabetes.csv")
```

```python
x = data.drop(["Outcome"], axis=1)
y = data["Outcome"]
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.25)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
x_train[:5]
```

```python
x_train.shape
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
```

```python
model=keras.Sequential([
    layers.Input(shape=(x_train.shape[1],),name='input_layer'), #input_layer has 8 neurons
    layers.Dense(16,activation='relu',name='hidden_layer_1'), #hidden_layer_1 has 16 neurons
    layers.Dense(32,activation='relu',name='hidden_layer_2'), #hidden_layer_2 has 32 neurons
    layers.Dense(1,activation='sigmoid',name='output_layer') #output_layer has just 1 neuron since it is a binary classifier
])
```

```python
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
model.fit(x_train,y_train,epochs=15,batch_size=32)
```

```python
y_pred=model.predict(x_test)
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score
```

```python
y_pred = (y_pred > 0.5).astype(int)
```

```python
accuracy=accuracy_score(y_test, y_pred)
print(accuracy)
```

```python
confusion_matrix(y_test, y_pred)
#                Predicted
#                  0    1
# Actual  0       TN   FP
# Acutual 1       FN   TP

#  100(TN) precited as non diabetic and they are also non diabetic
#  16(FP) predicted as diabetic but are actually non diabetic
#  27(FN) predicted as non diabetic but are actually diabetic
#  49(TP) predicted as diabetic and are also diabetic
```
